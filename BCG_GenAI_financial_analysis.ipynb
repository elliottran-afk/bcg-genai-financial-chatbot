```python
import pandas as pd
```

```python
pd.read_excel('BCG_GenAI_financial_data.xlsx').to_csv('financial_data.csv', index=False)

df = pd.read_csv('financial_data.csv')
df = df.sort_values(['Company', 'Year']).reset_index(drop=True)
df
```

```python
## Year-on-year growth
```

```python
metrics = ['Total Revenue', 'Net Income', 'Total Assets',
           'Total Liabilities', 'Cash Flow from Operating Activities']

for m in metrics:
    df[f'{m} Growth (%)'] = df.groupby('Company')[m].pct_change() * 100

df[['Company', 'Year'] + [f'{m} Growth (%)' for m in metrics]].round(1)
```

```python
df['Net Margin (%)'] = df['Net Income'] / df['Total Revenue'] * 100
df['Liabilities to Assets (%)'] = df['Total Liabilities'] / df['Total Assets'] * 100
df['Cash Flow to Net Income'] = df['Cash Flow from Operating Activities'] / df['Net Income']

df[['Company', 'Year', 'Net Margin (%)', 'Liabilities to Assets (%)', 'Cash Flow to Net Income']].round(2)
```

```python
print(df.groupby('Company')[['Total Revenue Growth (%)', 'Net Income Growth (%)', 'Net Margin (%)']].mean().round(1))

df.pivot(index='Year', columns='Company', values='Net Margin (%)').round(1)
```

```python
## Summary

**Method.** We took five figures from the 2023–2025 annual reports (10-K) that Microsoft, Apple and Tesla file with the SEC: total revenue, net income, total assets, total liabilities and cash flow from operating activities. We put them into a spreadsheet, converted it to CSV and analysed it with pandas. For each metric we calculated the year-on-year change by company, then added three ratios: net margin, liabilities to assets, and operating cash flow to net income.

**Microsoft** is the healthiest of the three. Revenue grew about 15% a year and profit grew faster than that, so net margin rose from 34% to 36%. Operating cash flow went up 35% in FY2024, and liabilities fell from 50% to 45% of assets. One thing to watch is that assets grew faster than revenue (partly the Activision Blizzard acquisition and data-centre spending), so Microsoft now makes less revenue per dollar of assets than it did in 2023.

**Apple** grows slowly, around 4% a year in revenue, but its profits are large and stable. Reported net income fell 3.4% in FY2024 and rose 19.5% in FY2025, but FY2024 included a one-off tax charge of about $10bn. Without it, profit grew roughly 7% in both years. Liabilities are about 80% of assets, which looks high. That's mainly because Apple spends heavily on buying back its own shares, which shrinks equity, so it isn't a sign of financial trouble.

**Tesla** is the least leveraged (liabilities have stayed at about 40% of assets), yet it has a real problem. Revenue was flat in 2024 and fell 2.9% in 2025, while net margin dropped from 15.5% to 7.3% and then to 4.0%. Some of the 2024 fall comes from a one-off tax benefit of about $5.9bn in 2023, but profit fell another 46% in 2025 with nothing like that to explain it. Operating cash flow has held at around $15bn, so the issue is profitability. Solvency isn't in question.

Overall, Microsoft and Apple sit well above Tesla on profitability. Apple's 2025 profit ($112bn) was bigger than Tesla's entire revenue ($95bn).

**Limitations.** Three years only gives two growth rates per company, so the trends rest on very little data. The financial years also end in different months (June for Microsoft, September for Apple, December for Tesla). One-off items distort the year-on-year figures too, as Apple and Tesla show, so the raw percentages need context.

**What this means for the chatbot.** Simple lookups like "What was Tesla's net income in 2025?" are easy to answer. But a bot that only reads numbers back would mislead people, for example by reporting Apple's 19.5% profit jump with no explanation. To be useful it needs to flag one-offs and the different financial years.
```
